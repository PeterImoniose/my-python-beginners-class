```python
p=[2345]
p1,p2,p3,p4=map(int,input("Enter your 4 digits pin separated by a comma(,):  ").split(","))
lock=int(f"{p1}{p2}{p3}{p4}")
for a in p:
    if lock == a:
        print("password correct")
        break
else:
    print("password incorrect")
```
