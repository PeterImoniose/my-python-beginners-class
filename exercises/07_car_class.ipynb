```python
class car:
    def __init__(self,brand,model,color,year,speed):
        self.brand=brand
        self.model=model
        self.color=color
        self.year=year
        self.speed=speed
    def start(self):
        print(f"{self.brand} {self.model} has started")
    def change_gear(self):
        option=input("Choose gear selection with [1,2,3,4,5,R]")
        if option == "1":
            print(f"Gear selection has changed to ONE {self.model}")
        elif option == "2":
            print(f"Gear selection has changed to TWO {self.model}")
        elif option == "3":
            print(f"Gear selection has changed to THREE {self.model}")
        elif option == "4":
            print(f"Gear selection has changed to FOUR {self.model}")
        elif option == "5":
            print(f"Gear selection has changed to FIVE {self.model}")
        elif option == "R":
            print("Car in reverse...")
    def acceleration(self):
        option=input("CHOOSE SPORT MODE[ON,OFF]")
        if option == "ON":
            print("WARNING!! SPORT MODE ENGAGED")
        elif option == "OFF":
            print("SPORT MODE OFF...")
    def stop(self):
        print("ENGINE OFF")
def main():
    car1=car('MERCEDES BENZ','AMG','BLACK',2023,'250MPH')
    while True:
        print("\n<<<WELCOME TO YOUR MERCEDES>>>\n")
        print("1.START")
        print("2.CHANGE GEAR")
        print("3.ACCELERATION")
        print("4.STOP")
        print("5.EXIT")
        choice=input("\nselect an option\n")
        if choice=="1":
            car1.start()
        elif choice=="2":
            car1.change_gear()
        elif choice=="3":
            car1.acceleration()
        elif choice=="4":
            car1.stop()
        elif choice=="5":
            print("THANK YOU FOR CHOOSING MERCEDES...")
            break
        else:
            print("INVALID SELECTION!!! PLEASE SELECT A VALID COMMAND...")
if __name__ == "__main__":
    main()
```
