```python
class car:
    def __init__(self,make,model,color,year,distance):
        self.make=make
        self.model=model
        self.color=color
        self.year=year
        self.distance=distance
    def start(self):
        print(f"{self.make} {self.model} {self.year} has started")
    def depr(self):
        year1= int(input("Year of purchase:  "))
        year2= int(input("Year of production:  "))
        if year1<year2:
            print("Error")
        else:
            cost=float(input("Whats the cost price?  $"))
            depreciation=int((cost/15)*(year1-year2))
            dep=int(cost/15)
            year3=int(year1-year2)
            cost2=int(cost-depreciation)
            print(f"The depreciation value of your {self.make} {self.model} {year2} after {year3} years is ${depreciation} and the current cost is ${cost2} at ${dep} per annum")
    def mileage(self):
        distance=float(input("What is the total distance covered?  "))
        fuel=float(input("What is the total quantity of fuel in litres used?  "))
        mile=distance/fuel
        print(f"Your fuel economy is {mile:.1f} miles per litre")
        last_oil_change_mileage=float(input("Enter the mileage at the last oil change: "))
        oil_change_interval=5000
        miles_since_last_change=distance-last_oil_change_mileage
        if miles_since_last_change>=oil_change_interval:
            print(f"It's time for an oil change for your {self.make} {self.model} {self.year}.")
        else:
            remaining_miles=oil_change_interval-miles_since_last_change
            print(f"You can wait for another {remaining_miles}miles for the next oil change.")
def main():
    car1=car('MERCEDES BENZ','AMG','BLACK',2023,'250MPH')
    while True:
        print("\n<<<WELCOME TO YOUR MERCEDES>>>\n")
        print("1.START")
        print("2.depreciation")
        print("3.Mileage")
        print("4.STOP")
        choice=input("\nselect an option\n")
        if choice=="1":
            car1.start()
        elif choice=="2":
            car1.depr()
        elif choice=="3":
            car1.mileage()
        elif choice=="4":
            print("THANK YOU FOR CHOOSING MERCEDES...")
            break
        else:
            print("INVALID SELECTION!!! PLEASE SELECT A VALID COMMAND...")
if __name__ == "__main__":
    main()
```
