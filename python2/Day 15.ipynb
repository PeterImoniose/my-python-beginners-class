```python
mycursor.execute("use kjprison")
mycursor.execute("create table prisoner_record(prisoner_id int primary key not null,name varchar(255) not null,age int not null,size float not null,dob date not null,visitors_name varchar(255) not null,reason_conviction varchar(255) not null,year varchar(8) not null,allergies varchar(45) not null,cell_no int not null)")
mycursor.execute("show columns from prisoner_record")
for x in mycursor:
    print(x)
```

```python
mycursor.execute("use kjprison")
def prison_record():
    prisoner_id = int(input("Enter your prisoner ID number:"))
    name = input("Enter Prisoner name:")
    age = int(input("Enter prisoner age:"))
    size = float(input("Enter Prisoner weight in kg:"))
    dob = input("Enter date of birth:")
    visitors_name = input("Enter visitor name:")
    reason_conviction = input("Enter the reason for prisoner conviction. Must be less than 50 words:")
    year = input("Enter the year of priner conviction:")
    allergies = input("Enter Prisoner Allergies:")
    cell_no = int(input("Enter prisoner cell number"))
    
    query = "insert into prisoner_record(prisoner_id,name,age,size,dob,visitors_name,reason_conviction,year,allergies,cell_no) values(%s,%s,%s,%s,%s,%s,%s,%s,%s,%s)"
    mycursor.execute(query,(prisoner_id,name,age,size,dob,visitors_name,reason_conviction,year,allergies,cell_no))
    mydb.commit()
    print("PRISONER RECORD SAVED SUCCESSFULLY")
```

```python
def search_prisoner_record():
    prisoner_id = int(input("Enter your prisoner ID number:"))
    query = 'select * from prisoner_record where prisoner_id = "%s"'
    mycursor.execute(query,(prisoner_id,))
    result = mycursor.fetchall()
    if len(result) == 0:
        print("Record not found...")
    else:
        for row in result:
            print(row)
def display_prisoners():
    mycursor.execute("select * from prisoner_record")
    for x in mycursor:
        print(x)
def delete_prisoner_record():
    prisoner_id=int(input("Enter Prisoner ID_NUMBER:"))
    query="delete from prisoner_record where prisoner_id = %s"
    mycursor.execute(query,(prisoner_id))
    mydb.commit()

def main():
    print("_____KUJE PRISONER INFORMATION_____")
    print("1.PRISONER_RECORD")
    print("2.SEARCH_PRISONER")
    print("3.Display_Prisoners")
    print("4.Delete_Prisoner_record")
    print("5.SIGN OUT")
    while True:
        option=input("SELECT OPERATION:")
        if option == "1":
            prison_record()
        elif option == "2":
            search_prisoner_record()
        elif option == "3":
            display_prisoners()
        elif option == "4":
            delete_prisoner_record()
        elif option == "5":
            print("LOGOUT..")
            break
        else:
            print("Invalid Option")
if __name__=="__main__":
    main()
```
