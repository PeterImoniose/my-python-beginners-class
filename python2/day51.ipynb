```python
## program a software that uses students subjects to look for the average of the scores and
## give a score point.
```

```python
print("check your grade!!!")
subject1=float(input("what is your mathematics score? "))
subject2=float(input("what is your Science score? "))
subject3=float(input("what is your Englsih score? "))
subject4=float(input("what is your Agricultural science score? "))
subject5=float(input("what is your Social Studies score? "))
subject6=float(input("what is your Literature score? "))
subject7=float(input("what is your Economics score? "))
import statistics as stat
avg=round(stat.mean([subject1,subject2,subject3,subject4,subject5,subject6,subject7]),2)
print(" ")
print(f"your average is {avg})
if avg>=75 and avg<=100:
    print("A")
elif avg>=60 and avg<75:
    print("B")
elif avg>=50 and avg <60:
    print("C")
elif avg>=45 and avg < 50:
    print("D")
else:
    print("F")
```

###### for loop

```python
name="Earlycode"
for i in name:
    print(i)
print(name)
```

```python
fruits=["banana","mango","pear","watermelon","orange","strawberry"]
for gift in fruits:
    print(gift)
print(fruits)
```

```python
word2="python"
for char in word2:
    print(char*4)
```

```python
## for loop(conditional statement)
word4="kelvin"
user=input("Enter the character name:")
for dave in word4:
    if user == dave:
        print("Character found...")
        break
else:
        print("character not found.")
```

```python
b=[2345]
p1,p2,p3,p4=map(int(s) for s in input("enter your password, seperate by a comma(,): ").split(","))
lock=int(f"{p1}{p2}{p3}{p4}")
print(lock)
for a in b:
    if lock == a:
        print("password correct")
        break
else:
    print("password incorrect")
```
