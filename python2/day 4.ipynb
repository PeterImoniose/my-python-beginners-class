## STATISTICS

```python
import statistics as stat
print(dir(stat))
```

```python
num1=[90,89,90,100,150,800,90]
print(stat.mean(num1))
print(stat.median(num1))
print(stat.mode(num1))
print(stat.quantiles(num1))
print(num1)
```
