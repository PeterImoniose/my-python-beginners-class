```python
## Exercise
## using function, write a program to calculate the total cost of a customer purhase allocated to each regions.
## including discounts and sales tax
## if the customer purchase is above 500000 you give a discount
## south south 15%
## south east 10%
## south west 8%
## north east 12%
## north west 11%
## north Central 6%
## The vat is slated at 3%
```

```python
def customer():
    purchaseamont =int(input("enter customer purchase amount"))
    region=input("Entetr customer region")
    if purchaseamont >= 500000 and region == "south south":
        print(f'''______CUSTOMER RECEIPT_____
        "purchase amount": {purchaseamont}
        "Discount":{purchaseamont * 0.15}
        "Tax":{purchaseamont *0.035}
        Total Amount Including Vat minus discount: {purchaseamont+(purchaseamont+0.035)-(purchaseamont*0.15)}
        ''')
    elif purchaseamont >= 500000 and region == "south east":
        print(f'''______CUSTOMER RECEIPT_____
        "purchase amount": {purchaseamont}
        "Discount":{purchaseamont * 0.1}
        "Tax":{purchaseamont *0.035}
        Total Amount Including Vat minus discount: {purchaseamont+(purchaseamont+0.035)-(purchaseamont*0.15)}
        ''')
    elif purchaseamont >= 500000 and region == "south west":
        print(f'''______CUSTOMER RECEIPT_____
        "purchase amount": {purchaseamont}
        "Discount":{purchaseamont * 0.08}
        "Tax":{purchaseamont *0.035}
        Total Amount Including Vat minus discount: {purchaseamont+(purchaseamont+0.035)-(purchaseamont*0.15)}
        ''')
    elif purchaseamont >= 500000 and region == "north east":
        print(f'''______CUSTOMER RECEIPT_____
        "purchase amount": {purchaseamont}
        "Discount":{purchaseamont * 0.12}
        "Tax":{purchaseamont *0.035}
        Total Amount Including Vat minus discount: {purchaseamont+(purchaseamont+0.035)-(purchaseamont*0.15)}
        ''')
    elif purchaseamont >= 500000 and region == "north west":
        print(f'''______CUSTOMER RECEIPT_____
        "purchase amount": {purchaseamont}
        "Discount":{purchaseamont * 0.11}
        "Tax":{purchaseamont *0.035}
        Total Amount Including Vat minus discount: {purchaseamont+(purchaseamont+0.035)-(purchaseamont*0.15)}
        ''')
    elif purchaseamont >= 500000 and region == "north centra":
        print(f'''______CUSTOMER RECEIPT_____
        "purchase amount": {purchaseamont}
        "Discount":{purchaseamont * 0.06}
        "Tax":{purchaseamont *0.035}
        Total Amount Including Vat minus discount: {purchaseamont+(purchaseamont+0.035)-(purchaseamont*0.15)}
        ''')
    else:
        print(f'''______CUSTOMER RECEIPT_____
        "purchase amount": {purchaseamont}
        "Tax":{purchaseamont *0.035}
        Total Amount Including Vat: {purchaseamont+(purchaseamont+0.035)}
        ''')
customer()
## USING FUNCTION AS A MODULE
```

```python
## MAP, FILTER LAMBDA
##### The map func executes a specific function on members of an iterable object. Each member is passed into the specified function as an argument.
##### The filter function filters the members of an iterable object if it 
```

#### LAMBDA

```python
def add(x,y):
    sum=x-y
    return sum
add(9,10)
```

```python
(lambda x,y,z:x + y-z)(9,10,8)
```

```python
## conditional statement for lambda
correct_p="Earlycode@1"
np=input("what is your password")
check_password=(lambda input_password:np ==correct_p)
print("Login Successfull" if check_password(np) else "Denied Access")
```

```python
## Map function with lambda.
numbers =[2,4,10,7,9,89,30,40]
square_number=list(map(lambda x:x**2,numbers))
print("Squared numbers", square_number)
```

```python
## Filter Function
num2=[2,3,4,6,7,9,12,23,46]
ylist=list(filter(lambda x:x%2 ==0, num2))
print("Even Numbers", ylist)
```

```python
## A PROGRAM TO DETERMINE THE CURRENC EXCHANGE.
def exchange (amount,exchange_rate):
    amount_received=amount*exchange_rate
    return amount_received
amount=float(input("What is the amount to be exchanged? "))
exchange_rate=float(input("What is the exchange rate? "))
result=exchange(amount,exchange_rate)
print("You will receive", result, "units of domestic currency after conversion")
```
