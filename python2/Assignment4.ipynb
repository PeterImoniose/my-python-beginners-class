```python
print("Election age check")
print("Determine your eligibility")
print("Note!!! Only 18+ are allowed to vote")
print(" ")
year=int(input("what is your year of birth:-"))
if year > 2024:
    print("error")
else:
    age=(2024-year)
    print(f"you are {age} years old")
    if age >= 18:
        print(f"You are Eligible")
    else:
        print("not Eligible")
```
