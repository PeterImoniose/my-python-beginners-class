#### ADDING TO CLASS IS KNOWN AS METHOD.

```python
class person:
    attrl="sam"
    attrl2="Male"
    attrl3=30
    def __init__(self):
        self.attrl
        self.attrl2
        self.attrl3
    def details(self):
        print(f"""name:{self.attrl} gender:{self.attrl2} age:{self.attrl3}""")
    def action(self):
        print(f"My name is {self.attrl} and my age is {self.attrl3} and i am a {self.attrl2}")
human2=person()
```

```python
human2.details()
```

```python
human2.action()
```
