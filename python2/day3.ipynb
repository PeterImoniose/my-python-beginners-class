# ITERABLE OBJECTS
##### iterable objects are capable of holding multiple value with indexes or keys pointing 

#### Turple
#### Turple uses index to locate their values
#### Turple is unchangeable
#### Turple uses this bracket ().

```python
num1=(20,25,90,100,260,90.0,"Turple",98)
print(type(num1))
print(len(num1))
print(num1)
```

```python
## INDEXING
### forward index
print(num1[6])
print(num1[7])
print(num1[4])
print(num1[0])
```

```python
## Backward index
print(num1[-3])
```

```python
## Range Index
print(num1[1:6])
```

```python
## Forward range index
print(num1[1:6])    # +1
```

```python
## full forward index
print(num1[0:8])
```

```python
## Backward range Index
print(num1[-4::])    # -1
```

```python
## Dictionary
## Dictionary is change-able
## Dictionary uses the curly bracket {}
## Dictionary uses key to access the value
```

```python
human={"Name":"Amira Joseph",
      "Marital_Status":"Married",
      "Age":30,
      "Occupation":"Data Scientist",
      "Country":"Nigeria",
      "State of origin":"Imo state",
      "DOB":"1998-09-20"}
print(type(human))
print(len(human))
print(human)
```

```python
## assign a New Value to a dictionary
human["Occupation"]="Data Analyst"
```

```python
print(human)
```

```python
del human["Occupation"]
```

```python
print(human)
```

```python
human["height"]=1.97
```

```python
## Dictionary compound.
student_grades={
    "Alice":{"math":90,"science":85,"english":100},
    "Bob":{"math":89,"science":50,"english":89},
    "Ade":{"math":90,"science":40,"english":95},
    "Charlie":{"math":89,"science":85,"english":45},
    "Felix":{"math":97,"science":67,"english":100}
}
print("Alice Grades")
print("math",student_grades["Alice"]["math"])
print("Science",student_grades["Alice"]["science"])
print("english",student_grades["Alice"]["english"])
```
