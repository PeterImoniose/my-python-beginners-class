```python
## Exercise
## using function, write a program to calculate the total cost of a customer purchase allocated to each regions.
## including discounts and sales tax
## if the customer purchase is above 500000 you give a discount
## south south 15%
## south east 10%
## south west 8%
## north east 12%
## north west 11%
## north Central 6%
## The vat is slated at 3%
```

```python
def customer():
    purchase_amount = int(input("Enter customer purchase amount: "))
    region = input("Enter customer region: ").lower()
    discounts = {"south south": 0.15, "south east": 0.10, "south west": 0.08,
                 "north east": 0.12, "north west": 0.11, "north central": 0.06}
    vat_rate = 0.03
    discount = purchase_amount * discounts.get(region, 0) if purchase_amount >= 500000 else 0
    tax = purchase_amount * vat_rate
    print(f'''______CUSTOMER RECEIPT_____
    Purchase amount: {purchase_amount:,.2f}
    Discount: {discount:,.2f}
    VAT (3%): {tax:,.2f}
    Total amount including VAT minus discount: {purchase_amount + tax - discount:,.2f}
    ''')
customer()
```

```python
## MAP, FILTER LAMBDA
##### The map func executes a specific function on members of an iterable object. Each member is passed into the specified function as an argument.
##### The filter function filters the members of an iterable object if it
```

#### LAMBDA

```python
def add(x,y):
    sum=x+y
    return sum
add(9,10)
```

```python
(lambda x,y,z:x + y-z)(9,10,8)
```

```python
## conditional statement for lambda
correct_p="Earlycode@1"
np=input("what is your password")
check_password=(lambda input_password:np ==correct_p)
print("Login Successful" if check_password(np) else "Denied Access")
```

```python
## Map function with lambda.
numbers =[2,4,10,7,9,89,30,40]
square_number=list(map(lambda x:x**2,numbers))
print("Squared numbers", square_number)
```

```python
## Filter Function
num2=[2,3,4,6,7,9,12,23,46]
ylist=list(filter(lambda x:x%2 ==0, num2))
print("Even Numbers", ylist)
```

```python
## A PROGRAM TO DETERMINE THE CURRENCY EXCHANGE.
def exchange (amount,exchange_rate):
    amount_received=amount*exchange_rate
    return amount_received
amount=float(input("What is the amount to be exchanged? "))
exchange_rate=float(input("What is the exchange rate? "))
result=exchange(amount,exchange_rate)
print("You will receive", result, "units of domestic currency after conversion")
```
