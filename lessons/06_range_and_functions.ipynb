```python
score1=0
score2=0
import random
for q in range(1,11):
    dice1=(random.randint(1,6)+random.randint(1,6))
    dice2=(random.randint(1,6)+random.randint(1,6))

    if dice1>dice2:
        score1+=5
        print("Player one wins!!!")
        print(f"Player One score is:{dice1}")
        print(f"Player Two score is:{dice2}")
    elif dice2>dice1:
        score2+=5
        print("Player two wins!!!")
        print(f"Player Two score is:{dice2}")
        print(f"Player One score is:{dice1}")
    elif dice2==dice1:
        score2+=1
        score1+=1
        print("Draw")
        print(f"Player One score is:{dice1}")
        print(f"Player Two score is:{dice2}")
if score1>score2:
    print(f"player one is the overall winner by {score1} point")
elif score2>score1:
    print(f"player two is the overall winner by {score2} point")
else:
    print("Draw")
```

```python
def vote(name,age,gender):
    print(f"""____Citizen Details___
    "Name":{name}
    "Age":{age}
    "gender":{gender}""")
    if age>=18 and gender=="Male":
        print("You are eligible to vote")
    else:
        print("not eligible")
vote("Sam David", 25,"Male")
```

```python
def vote(name,age=18,height=1.97):
    print(f"""____Citizen Details___
    "Name":{name}
    "Age":{age}
    "Height":{height}""")
    if age>=18 and height=="1.97":
        print("You are eligible to vote")
    else:
        print("not eligible")
vote("Sam David", 20,1.80)
```
